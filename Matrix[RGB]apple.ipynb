**pengambilan matriks gambar RGB menggunakan Python**
===
---



mengambil data gambar menggyunakan `pillow library`

untuk menampilkan gambar menggunakan `matplotlib.pyplot`

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

#open image
img = np.array(Image.open("Red_apple.png"))

plt.figure(figsize=(6,6))
plt.imshow(img)
plt.axis('off') # mematikan tanda sumbu
```

```python
img_R, img_G, img_B = img.copy(), img.copy(), img.copy()
img_R [:,:,(1, 2)] =0 # buang value 0 dari GB
img_G [:,:,(0, 2)] =0 # buang value 0 dari RB
img_B [:,:,(0, 1)] =0 # buang value 0 dari RG

# fig visual all
fig, axes = plt.subplots(1, 4, figsize=(20, 5)) # (1, 4) tata letak grid subplot

#  gambar asli
axes[0].imshow(img)
axes[0].set_title('Original Image') # judul untuk subplot
axes[0].axis('off') 

#  gambar channel merah
axes[1].imshow(img_R)
axes[1].set_title('Red Channel')
axes[1].axis('off') 

#  gambar channel hijau
axes[2].imshow(img_G)
axes[2].set_title('Green Channel')
axes[2].axis('off') 

#  gambar channel biru
axes[3].imshow(img_B)
axes[3].set_title('Blue Channel')
axes[3].axis('off') 

img_rgb = np.concatenate((img, img_R, img_G, img_B), axis= 1)


print('Img shape: ',img.shape)    # shape image

#  informasi matriks RGB
print("Gambar berhasil dibuka dan matriks RGB diperoleh.")
print("Bentuk (Shape) matriks RGB:", img.shape)
print("Tipe data (dtype) matriks:", img.dtype)

# sebagian kecil dari matriks RGB (5x5 piksel pertama)
# hanya 3 kolom pertama (RGB), mengabaikan channel Alpha jika ada
print("\nData matriks (5x5 piksel pertama - hanya RGB):\n", img[:5, :5, :3])


plt.show()
```
